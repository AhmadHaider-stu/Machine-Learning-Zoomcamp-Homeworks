```python
# Select the cars from Asia 
Asia_cars = df[df['origin']=='Asia']

# Select the needed columns
cols = ['vehicle_weight', 'model_year']

# Get the first 7 records from the dataframe
Asia_cars[cols].head(7)

# Get thr array 
X = Asia_cars[cols].head(7).values

# 5. X^T X
XTX = X.T @ X

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. w = (X^T X)^-1 X^T y
w = XTX_inv @ X.T @ y

# 9. Sum all elements of w
round(w.sum(),3)
#0.369 1pt
```
